# 01 · Clasificación con SVM: práctica guiada

Ejemplo propuesto: conjunto de datos Breast Cancer Wisconsin incluido en scikit-learn.

- **Pregunta:** ¿Puede una máquina de vectores de soporte (SVM) clasificar una muestra como benigna o maligna a partir de las características numéricas de sus núcleos celulares?
- **Unidad de análisis:** Una muestra de masa mamaria, representada por las mediciones de los núcleos celulares obtenidas de una imagen digitalizada de una aspiración con aguja fina. Cada fila corresponde a una muestra.
- **Target (variable objetivo):** La clase de la muestra: `0 = maligno` y `1 = benigno`, según la codificación de `load_breast_cancer` de scikit-learn.

> **Advertencia:** Este ejemplo tiene fines exclusivamente educativos y no constituye un diagnóstico médico ni sustituye la evaluación de un profesional de la salud.


```python
from pathlib import Path
import sys
ROOT = Path.cwd().resolve()
if not (ROOT / "src").exists():
    ROOT = ROOT.parent
assert (ROOT / "src/inf8239_u01").exists(), "Abra el notebook desde el proyecto"
sys.path.insert(0, str(ROOT / "src"))
reports = ROOT / "reports"
reports.mkdir(exist_ok=True)
print("Intérprete:", sys.executable)
from inf8239_u01.models import build_svm
```

```python
import pandas as pd
from sklearn.datasets import load_breast_cancer

bunch = load_breast_cancer(as_frame=True)
X = bunch.data.copy()
y = bunch.target.copy()

print(X.shape)
print(y.value_counts().sort_index())
X.head()
```

```python
audit = pd.DataFrame({
    "tipo": X.dtypes.astype(str),
    "ausentes": X.isna().sum(),
    "unicos": X.nunique()
})
print("Duplicados:", X.duplicated().sum())
print("Target dentro de X:", y.name in X.columns)
audit.head(10)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)
print(X_train.shape, X_test.shape)
print(y_train.value_counts(normalize=True).round(3))
print(y_test.value_counts(normalize=True).round(3))
```

```python
from sklearn.dummy import DummyClassifier
from sklearn.metrics import f1_score

dummy = DummyClassifier(strategy="most_frequent")
dummy.fit(X_train, y_train)
dummy_pred = dummy.predict(X_test)
print("F1 macro baseline:", f1_score(y_test, dummy_pred, average="macro"))
```

```python
svm = build_svm(probability=False)
svm.fit(X_train, y_train)
```

```python
from sklearn.model_selection import GridSearchCV, StratifiedKFold

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
search = GridSearchCV(
    estimator=svm,
    param_grid={
        "model__C": [0.1, 1, 10],
        "model__gamma": ["scale", 0.01, 0.1]
    },
    scoring="f1_macro", cv=cv, n_jobs=1, return_train_score=True
)
search.fit(X_train, y_train)
print(search.best_params_)
print("CV:", round(search.best_score_, 4))
```

```python
from sklearn.metrics import classification_report
cv_results = pd.DataFrame(search.cv_results_)
columns = ["param_model__C", "param_model__gamma",
           "mean_test_score", "std_test_score", "mean_fit_time"]
cv_results[columns].sort_values("mean_test_score", ascending=False).head(9)

best = search.best_estimator_
best_pred = best.predict(X_test)
print(classification_report(y_test, best_pred, digits=4))
```

```python
from pathlib import Path
import joblib

reports = ROOT / "reports"
reports.mkdir(parents=True, exist_ok=True)

cv_results[columns].to_csv(reports / "svm_cv_results.csv", index=False)
joblib.dump(best, reports / "svm_best.joblib")
```

## Evaluación final y análisis de errores
La selección de hiperparámetros usa únicamente el conjunto de entrenamiento y F1 macro en cinco pliegues estratificados. La partición de prueba es fija. Ya fue consultada en la práctica original: esta repetición no equivale a una validación externa nueva.
Para sensibilidad y precisión se considera positiva la malignidad (clase 0). El AUC usa el negativo de `decision_function`, que originalmente puntúa la clase 1. No se interpretan estos valores como probabilidades.


```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import (accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score, confusion_matrix, ConfusionMatrixDisplay, RocCurveDisplay)
rows = []
for name, model in [("Baseline", dummy), ("SVM base", svm), ("SVM ajustado", best)]:
    pred = model.predict(X_test)
    scores = (model.predict_proba(X_test)[:, 0] if name == "Baseline"
              else -model.decision_function(X_test))
    rows.append(dict(modelo=name, accuracy=accuracy_score(y_test, pred),
        f1_macro=f1_score(y_test, pred, average="macro"),
        precision_maligno=precision_score(y_test, pred, pos_label=0, zero_division=0),
        recall_maligno=recall_score(y_test, pred, pos_label=0),
        roc_auc=roc_auc_score((y_test == 0).astype(int), scores),
        malignos_como_benignos=int(((y_test == 0) & (pred == 1)).sum())))
results = pd.DataFrame(rows)
results.to_csv(reports / "resultados_finales.csv", index=False)
print(results.round(4).to_string(index=False))
audit.to_csv(reports / "auditoria.csv")
errors = X_test.loc[y_test != best_pred].copy()
errors["real"] = y_test.loc[errors.index]
errors["prediccion"] = best_pred[y_test != best_pred]
errors.to_csv(reports / "errores_svm.csv", index_label="fila")
fig, ax = plt.subplots(figsize=(5.6, 4.2))
ConfusionMatrixDisplay.from_predictions(y_test, best_pred, labels=[0,1],
    display_labels=["Maligno", "Benigno"], ax=ax, cmap="Blues", colorbar=False)
ax.set(title="SVM ajustado · 114 muestras", xlabel="Clase predicha", ylabel="Clase real")
fig.tight_layout()
fig.savefig(reports / "matriz_confusion.png", dpi=180)
plt.show()
fig, ax = plt.subplots(figsize=(5.6, 4.2))
RocCurveDisplay.from_predictions((y_test == 0).astype(int), -best.decision_function(X_test), ax=ax)
ax.plot([0,1], [0,1], "--", color="gray")
ax.set(title="ROC · malignidad como clase positiva", xlabel="Tasa de falsos positivos", ylabel="Sensibilidad")
fig.tight_layout()
fig.savefig(reports / "curva_roc.png", dpi=180)
plt.show()
summary = dict(filas=len(X), variables=X.shape[1], ausentes=int(X.isna().sum().sum()),
    duplicados=int(X.duplicated().sum()), train=len(X_train), test=len(X_test),
    best_params=search.best_params_, cv_f1=float(search.best_score_),
    confusion=confusion_matrix(y_test,best_pred,labels=[0,1]).tolist())
import json
(reports / "resumen.json").write_text(json.dumps(summary, indent=2), encoding="utf-8")
```
